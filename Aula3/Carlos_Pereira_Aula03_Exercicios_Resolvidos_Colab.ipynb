## Atividade — Aula 03

Resolução das questões sobre árvore de decisão, Random Forest e consumo de energia.

Para executar no Colab: **Ambiente de execução → Executar tudo**. O arquivo já contém a base de dados.


```python
#@title Carregar a base original incorporada ao arquivo
import base64
import io
import zlib
import numpy as np
import pandas as pd
from IPython.display import display
from sklearn.base import clone
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# Cópia compactada, sem alterações, do CSV original da atividade.
DADOS_CSV = ""
dados_brutos = pd.read_csv(io.BytesIO(zlib.decompress(base64.b64decode(DADOS_CSV))))
print(f"Base original carregada: {len(dados_brutos):,} registros e {len(dados_brutos.columns)} colunas.")
```

### Dados e modelos utilizados

A base ficou com 5.000 registros após a retirada de 50 linhas repetidas. Foram usados 4.000 registros no treinamento e 1.000 no teste. A classificação considera consumo alto acima de 400 kWh.

A árvore tem profundidade máxima 4. O Random Forest utiliza 100 árvores, com profundidade máxima 5. A célula abaixo prepara os dados e calcula os resultados usados nas respostas.


```python
# Preparação independente do notebook completo.
dados_limpos = dados_brutos.drop_duplicates().copy()
removidas = len(dados_brutos) - len(dados_limpos)
dados_limpos["timestamp"] = pd.to_datetime(dados_limpos["timestamp"], errors="coerce")
assert dados_limpos["consumo_kwh"].notna().all()
dados_limpos["consumo_alto"] = (dados_limpos["consumo_kwh"] > 400).astype(int)
numericas = ["producao_unidades", "temperatura_ambiente_c", "maquinas_ativas", "horas_operacao_turno"]
categoricas = ["turno"]
atributos = numericas + categoricas
Xc_treino, Xc_teste, yc_treino, yc_teste = train_test_split(
    dados_limpos[atributos], dados_limpos["consumo_alto"],
    test_size=0.20, random_state=42, stratify=dados_limpos["consumo_alto"])
preprocessamento = ColumnTransformer([
    ("num", SimpleImputer(strategy="median"), numericas),
    ("cat", OneHotEncoder(handle_unknown="ignore"), categoricas)
])
modelo_consumo = Pipeline([
    ("preparacao", clone(preprocessamento)),
    ("arvore", DecisionTreeClassifier(max_depth=4, random_state=42))
])
modelo_rf_consumo = Pipeline([
    ("preparacao", clone(preprocessamento)),
    ("floresta", RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42))
])
resultados = []
for nome, modelo in {"Árvore: consumo alto/baixo": modelo_consumo,
                     "Random Forest: consumo alto/baixo": modelo_rf_consumo}.items():
    modelo.fit(Xc_treino, yc_treino)
    previsao = modelo.predict(Xc_teste)
    tn, fp, fn, tp = confusion_matrix(yc_teste, previsao, labels=[0, 1]).ravel()
    resultados.append({"Modelo": nome,
        "Acurácia": accuracy_score(yc_teste, previsao),
        "Precisão": precision_score(yc_teste, previsao, zero_division=0),
        "Recall": recall_score(yc_teste, previsao, zero_division=0),
        "F1-score": f1_score(yc_teste, previsao, zero_division=0),
        "FP": int(fp), "FN": int(fn), "Erros": int(fp + fn)})
comparacao_consumo = pd.DataFrame(resultados)
print(f"Duplicatas removidas: {removidas} | Treino: {len(Xc_treino)} | Teste: {len(Xc_teste)}")
display(comparacao_consumo.round(4))
```

## 1. Respostas

### 1. Qual é a diferença entre `eficiente` e `consumo_alto`?

`eficiente` é a classificação fornecida no exemplo das 20 máquinas: 0 para ineficiente e 1 para eficiente. Já `consumo_alto` é definido pelo consumo de energia: acima de 400 kWh recebe 1; até 400 kWh recebe 0.

Consumo alto não significa necessariamente ineficiência. Uma máquina pode consumir mais energia porque está produzindo mais. São duas classificações diferentes.

### 2. Por que o teste passou de 1.010 para 1.000 registros?

O CSV tem 5.050 linhas, das quais 50 são repetidas. Com a retirada dessas linhas, restam 5.000 registros. Como 20% foram separados para teste, o conjunto ficou com 1.000 registros.

No slide, a tabela sem duplicatas foi guardada em outra variável, mas o restante do código continuou usando a tabela original. Isso explica os 1.010 registros apresentados na aula e a diferença nos resultados.

### 3. Por que excluir `consumo_kwh` das entradas?

Porque essa coluna já determina a resposta: basta verificar se o valor é maior que 400. Se ela entrasse no modelo, a avaliação daria uma impressão errada do desempenho, pois a resposta estaria nas próprias entradas.

Para prever o consumo alto ou baixo, as entradas são produção, temperatura, máquinas ativas, horas de operação e turno.

### 4. De onde vem a mediana utilizada no teste?

Ela vem dos dados de treinamento. O `SimpleImputer` calcula a mediana de cada coluna numérica durante o ajuste e usa esses valores para preencher os campos ausentes.

No teste, as medianas continuam sendo as do treino. Não são calculadas novamente com os dados que estão sendo avaliados.

### 5. `max_depth=2` em `plot_tree` reduz a complexidade do modelo?

Não. No `plot_tree`, esse parâmetro limita apenas o desenho. Os outros níveis continuam existindo no modelo.

Para limitar a profundidade no treinamento, o parâmetro precisa estar no classificador. A árvore deste exercício foi treinada com profundidade máxima 4.

### 6. O que acontece com um turno desconhecido?

Com `handle_unknown="ignore"`, o `OneHotEncoder` deixa os indicadores de turno com zero. A previsão é executada sem apresentar erro por causa da categoria nova.

Mesmo assim, o modelo não foi treinado com esse turno. Seria necessário avaliar os novos registros antes de confiar no resultado.

### 7. Uma temperatura acima de 50 °C deve ser removida automaticamente?

Não. Primeiro é preciso verificar se houve erro no sensor, na unidade ou no registro, ou se a temperatura realmente ocorreu. Excluir o valor sem essa conferência pode apagar uma situação importante do processo.

Nesta atividade, os valores foram mantidos para investigação.

### 8. O modelo com maior recall é sempre o melhor?

Não. Um recall maior significa que mais casos de consumo alto foram identificados, mas o modelo também pode gerar mais alertas incorretos.

A escolha depende do problema. Se deixar passar um consumo alto for mais prejudicial, o recall terá maior peso. Se os alertas desnecessários tiverem custo alto, também será importante observar a precisão e os falsos positivos.


## 2. Desafios práticos

### Desafio 1 — Custos dos erros

Foram comparadas três situações: custos iguais, falso negativo cinco vezes mais caro e falso positivo cinco vezes mais caro.

**Custo total = FP × custo do FP + FN × custo do FN**

Os valores servem para comparar os modelos; não são custos reais da indústria.


```python
cenarios_custos_finais = {
    "Custos iguais": (1, 1),
    "FN mais caro": (1, 5),
    "FP mais caro": (5, 1)
}
linhas_custos_finais = []
for cenario, (custo_fp, custo_fn) in cenarios_custos_finais.items():
    for _, resultado_modelo in comparacao_consumo.iterrows():
        linhas_custos_finais.append({
            "Cenário": cenario,
            "Modelo": resultado_modelo["Modelo"],
            "FP": int(resultado_modelo["FP"]),
            "FN": int(resultado_modelo["FN"]),
            "Custo de FP": custo_fp,
            "Custo de FN": custo_fn,
            "Custo total": int(resultado_modelo["FP"]) * custo_fp + int(resultado_modelo["FN"]) * custo_fn
        })
tabela_custos_finais = pd.DataFrame(linhas_custos_finais)
display(tabela_custos_finais)
display(tabela_custos_finais.pivot(index="Cenário", columns="Modelo", values="Custo total"))
```

A árvore teve 13 falsos positivos e 23 falsos negativos. O Random Forest teve 15 falsos positivos e 19 falsos negativos.

| Situação | Árvore | Random Forest | Menor custo |
|---|---:|---:|---|
| Custos iguais | 36 | 34 | Random Forest |
| FN custa cinco vezes mais | 128 | 110 | Random Forest |
| FP custa cinco vezes mais | 88 | 94 | Árvore |

Com o falso negativo mais caro, o Random Forest teve vantagem porque deixou passar menos casos de consumo alto. Quando o falso positivo ficou mais caro, a árvore teve o menor custo, pois gerou menos alertas incorretos.

Nesse teste, a escolha mudou de acordo com o custo do erro.


### Desafio 2 — Alterar uma variável por vez

O cenário inicial tem produção de 2.000 unidades, temperatura de 25 °C, seis máquinas ativas, sete horas de operação e turno da tarde.

Cada novo cenário muda apenas uma dessas informações. A comparação mostra a classe prevista e a probabilidade de consumo alto. Como são situações simuladas, não há uma medição de consumo para conferir se a previsão está correta.


```python
referencia_final = {
    "producao_unidades": 2000.0,
    "temperatura_ambiente_c": 25.0,
    "maquinas_ativas": 6.0,
    "horas_operacao_turno": 7.0,
    "turno": "tarde"
}
alteracoes_finais = {
    "Referência": {},
    "Produção de 1.000 unidades": {"producao_unidades": 1000.0},
    "Produção de 3.000 unidades": {"producao_unidades": 3000.0},
    "Temperatura de 35 °C": {"temperatura_ambiente_c": 35.0},
    "Nove máquinas ativas": {"maquinas_ativas": 9.0},
    "Quatro horas de operação": {"horas_operacao_turno": 4.0},
    "Turno da noite": {"turno": "noite"}
}
entradas_finais = pd.DataFrame(
    [{**referencia_final, **alteracao} for alteracao in alteracoes_finais.values()],
    index=list(alteracoes_finais)
)[atributos]
display(entradas_finais)
respostas_cenarios_finais = []
for nome, pipeline in {"Árvore": modelo_consumo, "Random Forest": modelo_rf_consumo}.items():
    classes = pipeline.predict(entradas_finais)
    probabilidades = pipeline.predict_proba(entradas_finais)[:, list(pipeline.classes_).index(1)]
    for cenario, classe, probabilidade in zip(entradas_finais.index, classes, probabilidades):
        respostas_cenarios_finais.append({"Cenário": cenario, "Modelo": nome,
            "Classe prevista": "Alto" if classe == 1 else "Baixo",
            "Probabilidade de Alto": round(float(probabilidade), 4)})
tabela_cenarios_finais = pd.DataFrame(respostas_cenarios_finais)
display(tabela_cenarios_finais)
for nome in ["Árvore", "Random Forest"]:
    recorte = tabela_cenarios_finais.loc[tabela_cenarios_finais["Modelo"] == nome].set_index("Cenário")
    classe_referencia = recorte.loc["Referência", "Classe prevista"]
    mudancas = recorte.loc[recorte["Classe prevista"] != classe_referencia].index.tolist()
    print(f"{nome}: referência = {classe_referencia}.")
    print("Cenários com mudança de classe:", "; ".join(mudancas) if mudancas else "Nenhum dos cenários testados.")
```

Na árvore, o cenário inicial foi classificado como alto. Com produção de 1.000 unidades, passou para baixo. Com 3.000 unidades, também ficou baixo. As outras alterações mantiveram a classe alta.

No Random Forest, o cenário inicial ficou baixo, com 46,4% de probabilidade para alto. Ao aumentar para nove máquinas ativas, passou para alto, com 88,2%. As outras alterações mantiveram a classe baixa.

O resultado para 3.000 unidades chama atenção: aumentar a produção não fez a previsão aumentar. A árvore segue os cortes aprendidos com a base, e não uma regra física de consumo. Esse caso precisaria ser investigado antes de usar o modelo na operação.


### Desafio 3 — Conferir as medianas

As medianas usadas no preenchimento estão em `statistics_`, no imputador numérico do pipeline.

A célula abaixo compara esses valores antes e depois de prever um cenário com temperatura ausente.


```python
imputador_final = modelo_consumo.named_steps["preparacao"].named_transformers_["num"]
medianas_antes_final = imputador_final.statistics_.copy()
cenario_ausente_final = pd.DataFrame([referencia_final])[atributos]
cenario_ausente_final.loc[0, "temperatura_ambiente_c"] = np.nan
classe_ausente_final = modelo_consumo.predict(cenario_ausente_final)[0]
medianas_depois_final = imputador_final.statistics_.copy()
display(pd.DataFrame({"Antes da previsão": medianas_antes_final,
    "Depois da previsão": medianas_depois_final}, index=numericas))
medianas_iguais_final = np.array_equal(medianas_antes_final, medianas_depois_final)
print("As medianas permaneceram iguais?", medianas_iguais_final)
print("Classe do cenário com temperatura ausente:", "Alto" if classe_ausente_final else "Baixo")
assert medianas_iguais_final
```

As medianas permaneceram iguais. O `predict()` usa os valores que já foram calculados no treinamento. A nova previsão não altera o imputador.

### Desafio 4 — Um ponto fora do boxplot é necessariamente um erro?

Não. Ele está distante da maior parte dos valores, mas pode ser uma medição correta. Um pico de produção ou uma mudança nas condições de operação pode aparecer dessa forma no gráfico. Por isso, o boxplot ajuda a localizar os registros que merecem conferência, mas não é motivo suficiente para excluí-los.


## 3. Conclusão

A árvore de decisão facilita a leitura das regras do modelo. No exemplo das 20 máquinas, o corte pela produção separou as classes e acertou os quatro casos de teste. Como são poucos registros, esse resultado ainda é limitado.

Na base de consumo, a árvore acertou 96,4% dos 1.000 registros de teste e o Random Forest acertou 96,6%. A diferença foi de dois acertos. A floresta deixou passar menos casos de consumo alto, enquanto a árvore gerou menos alertas incorretos.

A comparação de custos mostrou por que a acurácia não deve ser o único critério de escolha. Dependendo da consequência de cada erro, um modelo diferente pode ser mais adequado. Também ficou clara a importância de conferir as duplicatas e tratar os dados ausentes antes de interpretar o resultado.
