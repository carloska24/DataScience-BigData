## 📁 3. Teste de Leitura dos Dados da Aula 03
Vamos confirmar se o arquivo `consumo_energetico_alunos.csv` pode ser lido pelo Pandas:

````python
from pathlib import Path

caminho_csv = Path("consumo_energetico_alunos.csv")
if caminho_csv.is_file():
    df = pd.read_csv(caminho_csv)
    print(f"✅ Arquivo encontrado! Total de registros: {df.shape[0]} | Colunas: {df.shape[1]}\n")
    display(df.head())
else:
    print("⚠️ Arquivo consumo_energetico_alunos.csv não encontrado na pasta atual.")
````
