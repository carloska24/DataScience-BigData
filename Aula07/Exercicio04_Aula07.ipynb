**Tema:** Estrutura condicional para tomada de decisão e economia de custos.

### Enunciado original (PDF Aula 02, página 26)
> "Escreva um programa que pergunte o preço de três produtos e indique qual o produto com menor preço."

### Contexto e Aplicação
O programa avalia a cotação de três insumos ou produtos e orienta o setor de suprimentos a optar pelo item mais econômico.

```python
# 1. COTACÕES DOS PRODUTOS
produtos = {
    "Produto A (Fornecedor 1)": 45.90,
    "Produto B (Fornecedor 2)": 38.50,
    "Produto C (Fornecedor 3)": 52.00
}

# 2. LOCALIZAÇÃO DO PRODUTO MAIS ECONÔMICO
menor_produto = None
menor_valor = float("inf")

for nome, preco in produtos.items():
    if preco < menor_valor:
        menor_valor = preco
        menor_produto = nome

# 3. RELATÓRIO DE COMPRA
print("=== COTAÇÃO DE FORNECEDORES ===")
for nome, preco in produtos.items():
    print(f"{nome}: R$ {preco:.2f}")

print(f"\nRecomendação de Compra:")
print(f"-> Escolha: {menor_produto} pelo menor valor de R$ {menor_valor:.2f}.")
```

## Questões para análise

**a) O que acontece caso dois produtos empatem no menor valor?**
Se dois produtos tiverem o mesmo menor preço, a primeira condição avaliada será mantida ou podemos enriquecer o código para indicar empate entre os fornecedores.

**b) Qual o ganho financeiro obtido ao escolher a opção mais barata?**
Neste exemplo, a opção de R$ 38,50 frente à cotação mais cara (R$ 52,00) gera uma economia imediata de R$ 13,50 por unidade adquirida (25,96% de economia).

## Conclusão

A comparação condicional de preços automatiza decisões de compras corporativas, garantindo a seleção da proposta mais vantajosa para a indústria.
