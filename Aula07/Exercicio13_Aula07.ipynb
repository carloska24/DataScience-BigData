**Tema:** Operador lógico 'and' e verificação simultânea de credenciais.

### Enunciado original (PDF Aula 02, página 34)
> "Solicite o nome de usuário e a senha. Permita o acesso apenas quando o usuário for 'admin' e a senha for '1234'."

### Contexto e Aplicação
Sistema de controle de acesso que exige a conformidade conjunta de identificação de login e credencial de autenticação.

```python
# 1. FUNÇÃO DE AUTENTICAÇÃO
def autenticar(usuario: str, senha: str) -> bool:
    # Exige que ambas as condições sejam simultaneamente verdadeiras
    return (usuario == "admin") and (senha == "1234")

# 2. SIMULAÇÃO DE TENTATIVAS DE ACESSO
tentativas = [
    ("admin", "1234"),    # Ambas corretas (True and True -> True)
    ("admin", "9999"),    # Senha incorreta (True and False -> False)
    ("convidado", "1234"),# Usuário incorreto (False and True -> False)
    ("root", "toor")      # Ambas incorretas (False and False -> False)
]

print("=== SISTEMA DE LOGIN E SEGURANÇA ===")
for u, s in tentativas:
    sucesso = autenticar(u, s)
    if sucesso:
        print(f"Usuário: '{u:10s}' | Senha: '{s:5s}' -> ACESSO PERMITIDO! Bem-vindo ao painel.")
    else:
        print(f"Usuário: '{u:10s}' | Senha: '{s:5s}' -> ACESSO NEGADO! Credenciais inválidas.")
```

## Questões para análise

**a) Como funciona a tabela-verdade do operador `and`?**
A expressão `A and B` só retorna `True` quando **todas** as condições forem verdadeiras simultaneamente. Se qualquer uma falhar, o acesso é negado.

**b) O que garante a proteção das contas?**
Saber o nome do usuário não concede acesso, assim como ter a senha em uma conta incorreta também não autoriza o login.

## Conclusão

A autenticação baseada em `and` assegura a consistência das credenciais, sendo o pilar dos fluxos de autorização em sistemas computacionais.
