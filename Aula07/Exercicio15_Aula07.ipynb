**Tema:** Controle ambiental de precisão (salas limpas, estufas e galpões).

### Enunciado original (PDF Aula 02, página 34)
> "Solicite a temperatura e a umidade de um ambiente industrial. Exiba um alerta quando a temperatura ultrapassar 35 °C e a umidade estiver abaixo de 30%."

### Contexto e Aplicação
Dispara alerta severo quando o ar estiver simultaneamente muito quente (> 35 °C) e excessivamente seco (< 30%), condição propensa a descargas eletrostáticas e superaquecimento.

```python
# 1. FUNÇÃO DE MONITORAMENTO AMBIENTAL
def monitorar_clima_industrial(temperatura_c: float, umidade_pct: float):
    # Condição crítica: temperatura > 35 E umidade < 30
    alerta_critico = (temperatura_c > 35.0) and (umidade_pct < 30.0)
    
    if alerta_critico:
        diagnostico = "ALERTA CRÍTICO: Ambiente quente e seco! Risco de sobreaquecimento e eletricidade estática."
    else:
        diagnostico = "Ambiente dentro dos parâmetros normais de trabalho."
        
    return {
        "temperatura": temperatura_c,
        "umidade": umidade_pct,
        "alerta": alerta_critico,
        "diagnostico": diagnostico
    }

# 2. LEITURAS EM DIFERENTES GALPÕES
leituras = [
    (28.0, 55.0),  # Agradável (False and False)
    (38.0, 45.0),  # Quente, mas úmido (True and False)
    (30.0, 22.0),  # Frio/ameno, mas seco (False and True)
    (39.5, 18.0)   # Quente e muito seco (True and True -> ALERTA)
]

print("=== CONTROLE METEOROLÓGICO INDUSTRIAL (AND) ===")
for t, u in leituras:
    res = monitorar_clima_industrial(t, u)
    simbolo = "🔥 ALERTA" if res['alerta'] else "✅ NORMAL"
    print(f"Temp: {t:4.1f} °C | Umidade: {u:4.1f}% | {simbolo:10s} -> {res['diagnostico']}")
```

## Questões para análise

**a) Por que o operador `and` é exigido neste contexto?**
Porque uma temperatura de 36 °C com 60% de umidade não gera o mesmo risco eletrostático que um ambiente com apenas 20% de umidade. O alerta só deve soar quando ambas as condições de risco coexistirem.

**b) Qual ação corretiva deve ser tomada quando o alerta disparar?**
Acionamento de sistemas de umidificação industrial e ajuste dos condicionadores de ar para restabelecer a faixa segura de umidade (> 40%) e temperatura (< 25 °C).

## Conclusão

A verificação com `and` assegura diagnósticos de alta especificidade, prevenindo falsos alarmes em sistemas de controle predial e fabril.
