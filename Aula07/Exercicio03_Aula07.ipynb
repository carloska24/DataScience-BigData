**Tema:** Identificação de extremos (máximo e mínimo) em séries temporais automotivas.

### Enunciado original (PDF Aula 02, página 26)
> "Escreva um programa que leia valores médios de preços de um modelo de carro por 3 anos consecutivos e exiba o valor mais alto e mais baixo entre esses três anos."

### Contexto e Aplicação
Coleta os preços médios praticados por um modelo de veículo ao longo de um triênio (Ano 1, Ano 2 e Ano 3) e determina o preço ápice e o menor preço registrado.

```python
# 1. DADOS DE PREÇOS MÉDIOS DO VEÍCULO (3 ANOS)
anos = [2024, 2025, 2026]
precos = [85400.00, 92800.00, 88200.00]

# 2. DETERMINAÇÃO DO MAIOR E MENOR PREÇO VIA CONDICIONAIS
maior_preco = precos[0]
ano_maior = anos[0]
menor_preco = precos[0]
ano_menor = anos[0]

for ano, preco in zip(anos[1:], precos[1:]):
    if preco > maior_preco:
        maior_preco = preco
        ano_maior = ano
    if preco < menor_preco:
        menor_preco = preco
        ano_menor = ano

# 3. EXIBIÇÃO DOS RESULTADOS
print("=== HISTÓRICO DE PREÇOS MÉDIOS DO VEÍCULO ===")
for ano, p in zip(anos, precos):
    print(f"Ano {ano}: R$ {p:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))

print("\n=== EXTREMOS IDENTIFICADOS ===")
print(f"Maior preço médio: R$ {maior_preco:,.2f} (Ano {ano_maior})".replace(",", "X").replace(".", ",").replace("X", "."))
print(f"Menor preço médio: R$ {menor_preco:,.2f} (Ano {ano_menor})".replace(",", "X").replace(".", ",").replace("X", "."))
```

## Questões para análise

**a) Como identificar o maior e o menor valor em uma sequência de três entradas?**
Podemos inicializar as variáveis `maior` e `menor` com o primeiro valor recebido e, nas leituras seguintes, comparar condicionalmente se o novo valor supera o maior ou é inferior ao menor.

**b) É possível validar o resultado com as funções nativas `max()` e `min()`?**
Sim, em Python `max(precos)` e `min(precos)` retornam exatamente esses valores, confirmando a robustez da lógica algorítmica desenvolvida.

## Conclusão

A identificação dos valores extremos em séries consecutivas de preços permite analisar a curva de valorização ou depreciação de bens de capital e veículos na indústria.
