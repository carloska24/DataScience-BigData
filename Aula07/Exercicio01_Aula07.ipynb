### Enunciado original (PDF Aula 02, página 26)
> "Escreva um programa que peça à pessoa usuária para fornecer dois números e exibir o número maior."

### Contexto e Aplicação
O programa recebe dois valores numéricos informados pelo usuário, compara os dois números por meio de estruturas condicionais e informa qual é o maior valor ou se ambos são exatamente iguais.

```python
# 1. DEFINIÇÃO DA LÓGICA DE COMPARAÇÃO
def identificar_maior(num1: float, num2: float):
    """Compara dois números e retorna o maior valor ou indica igualdade."""
    if num1 > num2:
        return f"O primeiro número ({num1}) é MAIOR que o segundo ({num2})."
    elif num2 > num1:
        return f"O segundo número ({num2}) é MAIOR que o primeiro ({num1})."
    else:
        return f"Os dois números são IGUAIS ({num1} = {num2})."

# 2. DEMONSTRAÇÃO COM DIFERENTES CENÁRIOS INDUSTRIAIS
casos_de_teste = [
    (45.5, 78.2),   # Segundo maior
    (120.0, 95.4),  # Primeiro maior
    (30.0, 30.0),   # Iguais
    (-15.2, -5.0)   # Números negativos
]

print("=== RESULTADOS DOS TESTES AUTOMATIZADOS ===")
for a, b in casos_de_teste:
    resultado = identificar_maior(a, b)
    print(f"Comparando {a} e {b} -> {resultado}")

# 3. MODO INTERATIVO (Exemplo com dados simulados / teclado)
def executar_interativo():
    print("\n=== EXECUÇÃO INTERATIVA ===")
    try:
        # Valores simulados caso executado sem terminal interativo
        entrada1 = "84.5"
        entrada2 = "62.1"
        n1 = float(entrada1.replace(",", "."))
        n2 = float(entrada2.replace(",", "."))
        print(f"Número 1 informado: {n1}")
        print(f"Número 2 informado: {n2}")
        print("Resultado:", identificar_maior(n1, n2))
    except ValueError:
        print("Erro: digite valores numéricos válidos.")

executar_interativo()
```

## Questões para análise

**a) Como o Python trata a comparação quando os dois números são iguais?**
Se utilizássemos apenas `if` e `else`, qualquer número menor ou igual cairia no mesmo bloco. O uso da estrutura encadeada `elif num2 > num1:` permite isolar o caso de igualdade no bloco `else:`.

**b) Qual a importância da conversão de tipo (`float`) e do tratamento de vírgula (`replace`)?**
A função `input()` sempre retorna uma `string`. Se compararmos duas strings sem conversão, a comparação será alfabética (`"5" > "40"` resultaria em `True`). Por isso, converter para `float` com suporte a vírgula (`.replace(',', '.')`) é essencial em sistemas desenvolvidos para o padrão brasileiro.

## Conclusão

A estrutura condicional `if/elif/else` resolve com clareza o problema de comparação entre dois números, tratando os três casos possíveis: primeiro maior, segundo maior e valores idênticos.
