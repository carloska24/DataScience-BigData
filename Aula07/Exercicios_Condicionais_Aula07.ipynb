## Exercício 01 — Maior entre dois números (pág. 26)
> **Enunciado:** "Escreva um programa que peça à pessoa usuária para fornecer dois números e exibir o número maior."

```python
# 1. DEFINIÇÃO DA LÓGICA DE COMPARAÇÃO
def identificar_maior(num1: float, num2: float):
    """Compara dois números e retorna o maior valor ou indica igualdade."""
    if num1 > num2:
        return f"O primeiro número ({num1}) é MAIOR que o segundo ({num2})."
    elif num2 > num1:
        return f"O segundo número ({num2}) é MAIOR que o primeiro ({num1})."
    else:
        return f"Os dois números são IGUAIS ({num1} = {num2})."

# 2. DEMONSTRAÇÃO COM DIFERENTES CENÁRIOS INDUSTRIAIS
casos_de_teste = [
    (45.5, 78.2),   # Segundo maior
    (120.0, 95.4),  # Primeiro maior
    (30.0, 30.0),   # Iguais
    (-15.2, -5.0)   # Números negativos
]

print("=== RESULTADOS DOS TESTES AUTOMATIZADOS ===")
for a, b in casos_de_teste:
    resultado = identificar_maior(a, b)
    print(f"Comparando {a} e {b} -> {resultado}")

# 3. MODO INTERATIVO (Exemplo com dados simulados / teclado)
def executar_interativo():
    print("\n=== EXECUÇÃO INTERATIVA ===")
    try:
        # Valores simulados caso executado sem terminal interativo
        entrada1 = "84.5"
        entrada2 = "62.1"
        n1 = float(entrada1.replace(",", "."))
        n2 = float(entrada2.replace(",", "."))
        print(f"Número 1 informado: {n1}")
        print(f"Número 2 informado: {n2}")
        print("Resultado:", identificar_maior(n1, n2))
    except ValueError:
        print("Erro: digite valores numéricos válidos.")

executar_interativo()
```

### Análise e Conclusão (Exercício 01)

**a) Como o Python trata a comparação quando os dois números são iguais?**
Se utilizássemos apenas `if` e `else`, qualquer número menor ou igual cairia no mesmo bloco. O uso da estrutura encadeada `elif num2 > num1:` permite isolar o caso de igualdade no bloco `else:`.

**b) Qual a importância da conversão de tipo (`float`) e do tratamento de vírgula (`replace`)?**
A função `input()` sempre retorna uma `string`. Se compararmos duas strings sem conversão, a comparação será alfabética (`"5" > "40"` resultaria em `True`). Por isso, converter para `float` com suporte a vírgula (`.replace(',', '.')`) é essencial em sistemas desenvolvidos para o padrão brasileiro.

**Conclusão:** A estrutura condicional `if/elif/else` resolve com clareza o problema de comparação entre dois números, tratando os três casos possíveis: primeiro maior, segundo maior e valores idênticos.

## Exercício 02 — Percentual de crescimento da produção (pág. 26)
> **Enunciado:** "Escreva um programa que solicite o percentual de crescimento de produção de uma empresa e informe se houve um crescimento (porcentagem positiva) ou um decrescimento (porcentagem negativa)."

```python
# 1. FUNÇÃO DE CLASSIFICAÇÃO DA PRODUÇÃO
def avaliar_crescimento(percentual: float) -> str:
    """Classifica a variação da produção conforme o sinal do percentual."""
    if percentual > 0:
        return f"Crescimento na produção: aumento de +{percentual:.2f}%."
    elif percentual < 0:
        return f"Decrescimento na produção: queda de {percentual:.2f}%."
    else:
        return "Estabilidade na produção: variação nula (0.00%)."

# 2. TESTES COM DADOS INDUSTRIAIS HISTÓRICOS
historico_meses = {
    "Mês": ["Janeiro", "Fevereiro", "Março", "Abril", "Maio"],
    "Percentual (%)": [12.5, -4.8, 0.0, 8.3, -15.2]
}

print("=== ANÁLISE DE HISTÓRICO DE PRODUÇÃO ===")
for mes, pct in zip(historico_meses["Mês"], historico_meses["Percentual (%)"]):
    resultado = avaliar_crescimento(pct)
    print(f"{mes:10s} | Taxa: {pct:+6.2f}% -> {resultado}")
```

### Análise e Conclusão (Exercício 02)

**a) Como o sistema diferencia crescimento de decrescimento?**
Através da verificação lógica do sinal do valor: valores estritamente maiores que 0 (`> 0`) representam crescimento; valores estritamente menores que 0 (`< 0`) representam decrescimento; e o valor 0 representa estabilidade.

**b) Por que é recomendável prever o caso neutro (igual a zero)?**
Se houvesse apenas `if percentual > 0: ... else: ...`, um valor de 0% seria reportado erroneamente como decrescimento. A inclusão do caso zero garante precisão técnica ao relatório.

**Conclusão:** A análise condicional do percentual de crescimento permite que a diretoria fabril receba alertas automáticos e relatórios consistentes sobre a expansão ou retração das unidades produtivas.

## Exercício 03 — Preços médios de modelo automotivo por 3 anos (pág. 26)
> **Enunciado:** "Escreva um programa que leia valores médios de preços de um modelo de carro por 3 anos consecutivos e exiba o valor mais alto e mais baixo entre esses três anos."

```python
# 1. DADOS DE PREÇOS MÉDIOS DO VEÍCULO (3 ANOS)
anos = [2024, 2025, 2026]
precos = [85400.00, 92800.00, 88200.00]

# 2. DETERMINAÇÃO DO MAIOR E MENOR PREÇO VIA CONDICIONAIS
maior_preco = precos[0]
ano_maior = anos[0]
menor_preco = precos[0]
ano_menor = anos[0]

for ano, preco in zip(anos[1:], precos[1:]):
    if preco > maior_preco:
        maior_preco = preco
        ano_maior = ano
    if preco < menor_preco:
        menor_preco = preco
        ano_menor = ano

# 3. EXIBIÇÃO DOS RESULTADOS
print("=== HISTÓRICO DE PREÇOS MÉDIOS DO VEÍCULO ===")
for ano, p in zip(anos, precos):
    print(f"Ano {ano}: R$ {p:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))

print("\n=== EXTREMOS IDENTIFICADOS ===")
print(f"Maior preço médio: R$ {maior_preco:,.2f} (Ano {ano_maior})".replace(",", "X").replace(".", ",").replace("X", "."))
print(f"Menor preço médio: R$ {menor_preco:,.2f} (Ano {ano_menor})".replace(",", "X").replace(".", ",").replace("X", "."))
```

### Análise e Conclusão (Exercício 03)

**a) Como identificar o maior e o menor valor em uma sequência de três entradas?**
Podemos inicializar as variáveis `maior` e `menor` com o primeiro valor recebido e, nas leituras seguintes, comparar condicionalmente se o novo valor supera o maior ou é inferior ao menor.

**b) É possível validar o resultado com as funções nativas `max()` e `min()`?**
Sim, em Python `max(precos)` e `min(precos)` retornam exatamente esses valores, confirmando a robustez da lógica algorítmica desenvolvida.

**Conclusão:** A identificação dos valores extremos em séries consecutivas de preços permite analisar a curva de valorização ou depreciação de bens de capital e veículos na indústria.

## Exercício 04 — Menor preço entre três produtos (pág. 26)
> **Enunciado:** "Escreva um programa que pergunte o preço de três produtos e indique qual o produto com menor preço."

```python
# 1. COTACÕES DOS PRODUTOS
produtos = {
    "Produto A (Fornecedor 1)": 45.90,
    "Produto B (Fornecedor 2)": 38.50,
    "Produto C (Fornecedor 3)": 52.00
}

# 2. LOCALIZAÇÃO DO PRODUTO MAIS ECONÔMICO
menor_produto = None
menor_valor = float("inf")

for nome, preco in produtos.items():
    if preco < menor_valor:
        menor_valor = preco
        menor_produto = nome

# 3. RELATÓRIO DE COMPRA
print("=== COTAÇÃO DE FORNECEDORES ===")
for nome, preco in produtos.items():
    print(f"{nome}: R$ {preco:.2f}")

print(f"\nRecomendação de Compra:")
print(f"-> Escolha: {menor_produto} pelo menor valor de R$ {menor_valor:.2f}.")
```

**b) Qual o ganho financeiro obtido ao escolher a opção mais barata?**

## Exercício 05 — Cálculo de abastecimento e descontos em combustíveis (pág. 27)
> **Enunciado:** "Um estabelecimento está vendendo combustíveis com descontos variados:
- Etanol (E): R$ 1,70/litro. Se litros >= 15, desconto de 4% por litro. Senão, 3% por litro.
- Diesel (D): R$ 2,00/litro. Se litros >= 15, desconto de 5% por litro. Senão, 3% por litro.
- Gasolina (G): R$ 4,50/litro. Se litros >= 20, desconto de 3% por litro. Senão, sem desconto.
Escreva um programa que leia a quantidade de litros vendidos e o tipo de combustível (E, D, G) e calcule o valor a ser pago pelo cliente."

```python
# 1. FUNÇÃO DE CÁLCULO DE COMBUSTÍVEL
def calcular_combustivel(tipo: str, litros: float):
    tipo = tipo.upper().strip()
    
    if tipo == "E":
        nome = "Etanol"
        preco_litro = 1.70
        taxa_desconto = 0.04 if litros >= 15 else 0.03
    elif tipo == "D":
        nome = "Diesel"
        preco_litro = 2.00
        taxa_desconto = 0.05 if litros >= 15 else 0.03
    elif tipo == "G":
        nome = "Gasolina"
        preco_litro = 4.50
        taxa_desconto = 0.03 if litros >= 20 else 0.00
    else:
        return None, "Tipo de combustível inválido! Use E, D ou G."
    
    valor_bruto = preco_litro * litros
    valor_desconto = valor_bruto * taxa_desconto
    valor_final = valor_bruto - valor_desconto
    
    return {
        "tipo": tipo,
        "nome": nome,
        "litros": litros,
        "preco_litro": preco_litro,
        "taxa_desconto_pct": taxa_desconto * 100,
        "valor_bruto": valor_bruto,
        "valor_desconto": valor_desconto,
        "valor_final": valor_final
    }, None

# 2. SIMULAÇÃO DE CENÁRIOS
simulacoes = [
    ("E", 10.0),  # Etanol < 15L (3%)
    ("E", 25.0),  # Etanol >= 15L (4%)
    ("D", 12.0),  # Diesel < 15L (3%)
    ("D", 40.0),  # Diesel >= 15L (5%)
    ("G", 15.0),  # Gasolina < 20L (0%)
    ("G", 30.0),  # Gasolina >= 20L (3%)
]

print("=== RELATÓRIO DE ABASTECIMENTOS ===")
for t, l in simulacoes:
    res, err = calcular_combustivel(t, l)
    if res:
        print(f"Combustível: {res['nome']:8s} | Litros: {res['litros']:4.1f} L | "
              f"Bruto: R$ {res['valor_bruto']:6.2f} | Desc: {res['taxa_desconto_pct']:3.1f}% (R$ {res['valor_desconto']:5.2f}) | "
              f"Total a Pagar: R$ {res['valor_final']:6.2f}")
```

### Análise e Conclusão (Exercício 05)

**a) Qual a regra de desconto aplicada para a Gasolina?**
A gasolina só recebe desconto (de 3%) se o volume abastecido for de no mínimo 20 litros. Para volumes inferiores a 20 litros, não há desconto aplicado (0%).

**b) Como o cálculo do desconto foi estruturado?**
Conforme a dica do enunciado, o valor do desconto é calculado multiplicando-se o preço por litro pelo total de litros e pela taxa de desconto. O valor a ser pago é o valor bruto deduzido do desconto.

**Conclusão:** O algoritmo implementa com precisão as tabelas de tarifas e descontos escalonados por tipo de combustível e faixa de volume.

## Exercício 06 — Variação de vendas imobiliárias e diretrizes de gestão (pág. 28)
> **Enunciado:** "Em uma empresa de venda de imóveis você precisa criar um código que analise os dados de vendas anuais para ajudar a diretoria na tomada de decisão. O código precisa coletar os dados de quantidade de venda durante os anos de 2022 e 2023 e fazer um cálculo de variação percentual:
- Acima de 20%: bonificação para o time de vendas.
- Entre 2% e 20%: pequena bonificação para time de vendas.
- Entre 2% e -10%: planejamento de políticas de incentivo às vendas.
- Abaixo de -10%: corte de gastos."

```python
# 1. FUNÇÃO DE ANÁLISE DE VENDAS
def analisar_desempenho_imoveis(vendas_2022: int, vendas_2023: int):
    if vendas_2022 <= 0:
        return None, "Vendas de 2022 devem ser maiores que zero."
    
    variacao_pct = ((vendas_2023 - vendas_2022) / vendas_2022) * 100
    
    if variacao_pct > 20.0:
        sugestao = "Bonificação para o time de vendas."
    elif 2.0 <= variacao_pct <= 20.0:
        sugestao = "Pequena bonificação para time de vendas."
    elif -10.0 <= variacao_pct < 2.0:
        sugestao = "Planejamento de políticas de incentivo às vendas."
    else:
        sugestao = "Corte de gastos."
        
    return {
        "v2022": vendas_2022,
        "v2023": vendas_2023,
        "variacao_pct": variacao_pct,
        "sugestao": sugestao
    }, None

# 2. CENÁRIOS DE FILIAIS DA IMOBILIÁRIA
filiais = [
    ("Filial Norte", 100, 125),   # +25% (> 20%)
    ("Filial Sul", 200, 218),     # +9% (2% a 20%)
    ("Filial Leste", 150, 148),   # -1.33% (-10% a 2%)
    ("Filial Oeste", 180, 150)    # -16.67% (< -10%)
]

print("=== AVALIAÇÃO DE DESEMPENHO DAS FILIAIS ===")
for nome, v22, v23 in filiais:
    res, _ = analisar_desempenho_imoveis(v22, v23)
    print(f"{nome:14s} | 2022: {v22:3d} un | 2023: {v23:3d} un | Variação: {res['variacao_pct']:+6.2f}% | Decisão: {res['sugestao']}")
```

### Análise e Conclusão (Exercício 06)

**a) Qual a fórmula utilizada para a variação percentual?**
`variacao = ((vendas_2023 - vendas_2022) / vendas_2022) * 100`, que expressa o crescimento ou a queda relativa em relação à base de 2022.

**b) Quais faixas de ação foram contempladas?**
Quatro categorias exatas de governança corporativa: bonificação alta, bonificação moderada, necessidade de incentivo comercial e contingenciamento com corte de custos.

**Conclusão:** O modelo condicional automatiza a aplicação das políticas comerciais da empresa imobiliária, alinhando a remuneração e os planos de ação aos resultados concretos de vendas.

## Exercício 07 — Calculadora aritmética com tratamento de divisão por zero (pág. 29)
> **Enunciado:** "Desenvolva um programa que solicite dois números e uma operação matemática (+, -, * ou /). Utilize estruturas condicionais para identificar a operação e apresentar o resultado. Caso o usuário escolha uma divisão por zero, exiba uma mensagem de erro."

```python
# 1. FUNÇÃO DA CALCULADORA
def calcular(num1: float, num2: float, operacao: str):
    operacao = operacao.strip()
    
    if operacao == "+":
        return num1 + num2, None
    elif operacao == "-":
        return num1 - num2, None
    elif operacao == "*":
        return num1 * num2, None
    elif operacao == "/":
        if num2 == 0:
            return None, "ERRO: Divisão por zero não é permitida!"
        return num1 / num2, None
    else:
        return None, f"ERRO: Operação '{operacao}' inválida! Escolha +, -, * ou /."

# 2. TESTES DE TODAS AS OPERAÇÕES E CASOS ESPECIAIS
testes = [
    (15, 3, "+"),
    (20, 7, "-"),
    (6, 8, "*"),
    (48, 6, "/"),
    (10, 0, "/"),  # Divisão por zero
    (5, 2, "%")    # Operador desconhecido
]

print("=== DEMONSTRAÇÃO DA CALCULADORA ===")
for n1, n2, op in testes:
    res, err = calcular(n1, n2, op)
    if err:
        print(f"{n1} {op} {n2} -> {err}")
    else:
        print(f"{n1} {op} {n2} = {res:.2f}")
```

### Análise e Conclusão (Exercício 07)

**a) Por que o teste do divisor zero precisa ocorrer antes da divisão?**
Em Python, tentar calcular `x / 0` gera uma exceção `ZeroDivisionError` que interrompe o programa. Ao verificar com `if num2 == 0:` antes da operação, tratamos a falha de forma graciosa e amigável.

**b) Quais operadores são aceitos?**
Adição (`+`), subtração (`-`), multiplicação (`*`) e divisão (`/`). Operadores não previstos acionam o bloco final de erro de sintaxe.

**Conclusão:** A calculadora condicional demonstra a implementação correta de seletores de operações e a prevenção de erros críticos de divisão por zero.

## Exercício 08 — Descontos progressivos por faixa de valor de compra (pág. 30)
> **Enunciado:** "Uma loja oferece descontos de acordo com o valor total da compra. Solicite o nome do cliente e o valor da compra:
- Abaixo de R$ 100: 0%
- De R$ 100 até menos de R$ 500: 5%
- De R$ 500 até menos de R$ 1.000: 10%
- R$ 1.000 ou mais: 15%"

```python
# 1. FUNÇÃO DE CÁLCULO DE DESCONTO EM COMPRAS
def calcular_desconto_compra(nome: str, valor: float):
    if valor < 0:
        return None, "Valor da compra não pode ser negativo."
    
    if valor < 100.00:
        pct_desconto = 0.00
    elif valor < 500.00:
        pct_desconto = 0.05
    elif valor < 1000.00:
        pct_desconto = 0.10
    else:
        pct_desconto = 0.15
        
    valor_desconto = valor * pct_desconto
    valor_final = valor - valor_desconto
    
    return {
        "cliente": nome,
        "valor_original": valor,
        "pct_desconto": pct_desconto * 100,
        "valor_desconto": valor_desconto,
        "valor_final": valor_final
    }, None

# 2. CENÁRIOS DE CLIENTES EM CADA FAIXA
clientes_teste = [
    ("Ana Souza", 80.00),     # < 100 (0%)
    ("Bruno Lima", 250.00),   # 100 a < 500 (5%)
    ("Carla Dias", 750.00),   # 500 a < 1000 (10%)
    ("Daniel Silva", 1400.00) # >= 1000 (15%)
]

print("=== NOTA DE VENDA COM DESCONTO ===")
for c, v in clientes_teste:
    res, _ = calcular_desconto_compra(c, v)
    print(f"Cliente: {res['cliente']:13s} | Compra: R$ {res['valor_original']:7.2f} | "
          f"Desconto: {res['pct_desconto']:2.0f}% (R$ {res['valor_desconto']:6.2f}) | "
          f"Total a Pagar: R$ {res['valor_final']:7.2f}")
```

### Análise e Conclusão (Exercício 08)

**a) Como os limites de corte foram definidos no encadeamento de `if/elif`?**
Como a primeira condição checa `< 100`, quando o código atinge o `elif valor < 500:`, sabemos com certeza matemática que o valor já é maior ou igual a 100, dispensando redundâncias como `valor >= 100 and valor < 500`.

**b) Qual o benefício dessa política de descontos para a loja?**
Estimula o aumento do ticket médio, incentivando clientes próximos de R$ 500 ou R$ 1.000 a comprarem mais itens para alcançar a faixa seguinte de benefício.

**Conclusão:** A parametrização condicional de descontos progressivos oferece controle preciso de precificação e incentivo às vendas.

## Exercício 09 — Consumo energético e classificação de máquinas industriais (pág. 31)
> **Enunciado:** "Uma empresa deseja identificar o nível de consumo energético de suas máquinas. Solicite o nome da máquina, sua potência em kW e a quantidade de horas de funcionamento.
Calcule o consumo: Consumo (kWh) = Potência (kW) × Horas.
Classifique o resultado conforme faixas didáticas:
- Menos de 100 kWh: Baixo consumo.
- De 100 até menos de 300 kWh: Consumo moderado.
- 300 kWh ou mais: Alto consumo."

```python
# 1. FUNÇÃO DE CÁLCULO E CLASSIFICAÇÃO ENERGÉTICA
def classificar_consumo(nome: str, potencia_kw: float, horas: float):
    consumo_kwh = potencia_kw * horas
    
    if consumo_kwh < 100.0:
        categoria = "Baixo consumo"
    elif consumo_kwh < 300.0:
        categoria = "Consumo moderado"
    else:
        categoria = "Alto consumo"
        
    return {
        "maquina": nome,
        "potencia_kw": potencia_kw,
        "horas": horas,
        "consumo_kwh": consumo_kwh,
        "categoria": categoria
    }

# 2. AUDITORIA EM LINHA DE PRODUÇÃO
parque_maquinas = [
    ("Furadeira de Bancada", 1.5, 8.0),     # 12 kWh (Baixo)
    ("Torno Mecânico", 15.0, 12.0),         # 180 kWh (Moderado)
    ("Prensa Hidráulica", 45.0, 16.0),      # 720 kWh (Alto)
    ("Compressor Industrial", 25.0, 10.0)   # 250 kWh (Moderado)
]

print("=== AUDITORIA DE CONSUMO ENERGÉTICO INDUSTRIAL ===")
for m, p, h in parque_maquinas:
    res = classificar_consumo(m, p, h)
    print(f"Máquina: {res['maquina']:22s} | Potência: {res['potencia_kw']:4.1f} kW | "
          f"Horas: {res['horas']:4.1f} h | Consumo: {res['consumo_kwh']:6.1f} kWh | Faixa: {res['categoria']}")
```

### Análise e Conclusão (Exercício 09)

**a) Como se relaciona a potência do motor com o tempo de operação?**
A energia total consumida em quilowatt-hora (kWh) é diretamente proporcional tanto à potência de placa da máquina quanto ao total de horas em que ela permaneceu ligada.

**b) Quais máquinas requerem atenção prioritária de gestão energética?**
Equipamentos enquadrados em 'Alto consumo' (acima de 300 kWh), como prensas de grande porte, devem ser inspecionados quanto a desperdícios e operados preferencialmente fora dos horários de pico tarifário.

**Conclusão:** O monitoramento automatizado do consumo energético apoia estratégias de sustentabilidade (ESG) e redução de custos industriais.

## Exercício 10 — Avaliação de desempenho de vendas com validação de metas (pág. 32)
> **Enunciado:** "Uma empresa deseja analisar o desempenho mensal de seus vendedores. Desenvolva um programa que solicite o nome de um vendedor, sua meta mensal e o valor total vendido.
O programa deverá calcular o percentual de atingimento da meta e classificar o desempenho:
- Abaixo de 80%: Abaixo da meta
- 80% até menos de 100%: Próximo da meta
- 100% até menos de 120%: Meta atingida
- 120% ou mais: Meta superada
Valide se a meta é maior que zero e se as vendas não são negativas."

```python
# 1. FUNÇÃO COM VALIDAÇÃO E CLASSIFICAÇÃO DE METAS
def avaliar_vendedor(nome: str, meta: float, vendas: float):
    # Validações exigidas no enunciado
    if meta <= 0:
        return None, "ERRO: A meta mensal deve ser estritamente maior que zero!"
    if vendas < 0:
        return None, "ERRO: O total de vendas não pode ser negativo!"
        
    percentual = (vendas / meta) * 100
    
    if percentual < 80.0:
        situacao = "Abaixo da meta"
    elif percentual < 100.0:
        situacao = "Próximo da meta"
    elif percentual < 120.0:
        situacao = "Meta atingida"
    else:
        situacao = "Meta superada"
        
    return {
        "vendedor": nome,
        "meta": meta,
        "vendas": vendas,
        "percentual": percentual,
        "situacao": situacao
    }, None

# 2. AVALIAÇÃO DA EQUIPE COMERCIAL
equipe = [
    ("Lucas Mendes", 50000.00, 35000.00),   # 70% (Abaixo da meta)
    ("Mariana Costa", 60000.00, 54000.00),  # 90% (Próximo da meta)
    ("Ricardo Alves", 40000.00, 44000.00),  # 110% (Meta atingida)
    ("Juliana Prado", 50000.00, 65000.00),  # 130% (Meta superada)
    ("Teste Inválido", -1000.00, 5000.00)   # Meta <= 0 (Erro)
]

print("=== RELATÓRIO MENSAL DE PERFORMANCE DE VENDAS ===")
for nome, meta, vendas in equipe:
    res, err = avaliar_vendedor(nome, meta, vendas)
    if err:
        print(f"{nome:15s} | {err}")
    else:
        print(f"{res['vendedor']:15s} | Meta: R$ {res['meta']:8.2f} | "
              f"Vendas: R$ {res['vendas']:8.2f} | Atingimento: {res['percentual']:6.1f}% | "
              f"Classificação: {res['situacao']}")
```

### Análise e Conclusão (Exercício 10)

**a) Por que a validação é crucial antes do cálculo?**
Se a meta pudesse ser zero, o cálculo `(vendas / meta)` causaria um erro fatal de divisão por zero. A validação protege a integridade do sistema.

**b) Quais faixas definem o sucesso do vendedor?**
Atingimentos a partir de 100% caracterizam o cumprimento do objetivo comercial, com bonificação especial para desempenhos que superem 120%.

**Conclusão:** A validação prévia combinada com faixas claras de percentual produz um sistema confiável para cálculo de bônus e acompanhamento de metas.

## Exercício 11 — Monitoramento térmico de máquinas e alerta crítico (pág. 33)
> **Enunciado:** "Uma indústria utiliza sensores para monitorar a temperatura de suas máquinas. Desenvolva um programa que solicite o nome da máquina e sua temperatura atual em graus Celsius. Utilize estruturas condicionais para determinar o estado operacional:
- Abaixo de 40 °C: Operação normal
- De 40 °C até menos de 60 °C: Atenção
- De 60 °C até 80 °C: Alerta de aquecimento
- Acima de 80 °C: Temperatura crítica
Caso a temperatura seja crítica, apresente a mensagem: 'ALERTA: verificar imediatamente a máquina!'"

```python
# 1. FUNÇÃO DE MONITORAMENTO TÉRMICO
def monitorar_temperatura(nome_maquina: str, temp_c: float):
    if temp_c < 40.0:
        estado = "Operação normal"
        alerta_critico = False
    elif temp_c < 60.0:
        estado = "Atenção"
        alerta_critico = False
    elif temp_c <= 80.0:
        estado = "Alerta de aquecimento"
        alerta_critico = False
    else:
        estado = "Temperatura crítica"
        alerta_critico = True
        
    mensagem_alerta = "ALERTA: verificar imediatamente a máquina!" if alerta_critico else "Sem pendências de emergência."
    
    return {
        "maquina": nome_maquina,
        "temperatura": temp_c,
        "estado": estado,
        "critico": alerta_critico,
        "mensagem": mensagem_alerta
    }

# 2. LEITURA DOS SENSORES EM TEMPO REAL
sensores = [
    ("Bomba Centrífuga 01", 32.5),   # < 40
    ("Injetora Plástica 02", 48.0),  # 40 a < 60
    ("Extrusora 03", 72.0),          # 60 a 80
    ("Motor Principal 04", 94.5)     # > 80 (Crítica)
]

print("=== QUADRO DE CONTROLE DE TEMPERATURA FABRIL ===")
for maq, t in sensores:
    res = monitorar_temperatura(maq, t)
    print(f"Máquina: {res['maquina']:22s} | Temp: {res['temperatura']:5.1f} °C | Estado: {res['estado']:22s}")
    if res['critico']:
        print(f"   >>> {res['mensagem']}")
```

### Análise e Conclusão (Exercício 11)

**a) Como o limite crítico de 80 °C é tratado?**
Temperaturas de até 80 °C pertencem ao 'Alerta de aquecimento'. Valores estritamente maiores que 80 °C acionam a categoria 'Temperatura crítica' e a mensagem imperativa de verificação imediata.

**b) Qual o papel desse sistema na manutenção preditiva?**
Identificar anomalias nas faixas de atenção e aquecimento evita paradas inesperadas de linha e prolonga a vida útil dos equipamentos industriais.

**Conclusão:** O monitoramento térmico por faixas permite uma abordagem preventiva escalonada, evitando que temperaturas críticas danifiquem os motores.

## Exercício 12 — Alerta preditivo de temperatura ou vibração (Operador OR) (pág. 34)
> **Enunciado:** "Um sistema monitora a temperatura e a vibração de uma máquina industrial. Criar um programa para permitir que o usuário digite a temperatura e o sistema emita um alerta quando a temperatura ultrapassar 80 °C ou a vibração ultrapassar 15 mm/s."

```python
# 1. FUNÇÃO DE MONITORAMENTO COM OPERADOR OR
def verificar_maquina_temp_vib(temp_c: float, vibracao_mms: float):
    # Condição lógica: temperatura > 80 OU vibração > 15
    emite_alerta = (temp_c > 80.0) or (vibracao_mms > 15.0)
    
    if emite_alerta:
        status = "ALERTA EMITIDO: Condições fora dos limites de segurança!"
    else:
        status = "Operação Normal: Sensores dentro dos padrões tolerados."
        
    return {
        "temperatura": temp_c,
        "vibracao": vibracao_mms,
        "alerta": emite_alerta,
        "status": status
    }

# 2. TESTES DA TABELA-VERDADE DO OPERADOR OR
cenarios = [
    (65.0, 8.0),   # Ambas normais (False or False -> False)
    (88.0, 10.0),  # Temp alta, Vib normal (True or False -> True)
    (70.0, 18.5),  # Temp normal, Vib alta (False or True -> True)
    (92.0, 22.0)   # Ambas altas (True or True -> True)
]

print("=== ANÁLISE DE SENSORES CONJUNTOS (OR) ===")
for t, v in cenarios:
    r = verificar_maquina_temp_vib(t, v)
    simbolo = "🔴" if r["alerta"] else "🟢"
    print(f"{simbolo} Temp: {t:4.1f} °C | Vibração: {v:4.1f} mm/s -> {r['status']}")
```

### Análise e Conclusão (Exercício 12)

**a) Como funciona a tabela-verdade do operador `or`?**
A expressão `A or B` resulta em `True` se pelo menos uma das condições for verdadeira. Só será `False` se tanto a temperatura for <= 80 °C quanto a vibração for <= 15 mm/s.

**b) Por que o `or` é a escolha correta para este monitoramento?**
Porque tanto o superaquecimento isolado quanto o desbalanceamento mecânico com alta vibração podem destruir o equipamento de forma independente.

**Conclusão:** O operador `or` garante máxima segurança operacional, pois qualquer anomalia física detectada é suficiente para alertar a equipe técnica.

## Exercício 13 — Controle de acesso e autenticação de usuários (Operador AND) (pág. 34)
> **Enunciado:** "Solicite o nome de usuário e a senha. Permita o acesso apenas quando o usuário for 'admin' e a senha for '1234'."

```python
# 1. FUNÇÃO DE AUTENTICAÇÃO
def autenticar(usuario: str, senha: str) -> bool:
    # Exige que ambas as condições sejam simultaneamente verdadeiras
    return (usuario == "admin") and (senha == "1234")

# 2. SIMULAÇÃO DE TENTATIVAS DE ACESSO
tentativas = [
    ("admin", "1234"),    # Ambas corretas (True and True -> True)
    ("admin", "9999"),    # Senha incorreta (True and False -> False)
    ("convidado", "1234"),# Usuário incorreto (False and True -> False)
    ("root", "toor")      # Ambas incorretas (False and False -> False)
]

print("=== SISTEMA DE LOGIN E SEGURANÇA ===")
for u, s in tentativas:
    sucesso = autenticar(u, s)
    if sucesso:
        print(f"Usuário: '{u:10s}' | Senha: '{s:5s}' -> ACESSO PERMITIDO! Bem-vindo ao painel.")
    else:
        print(f"Usuário: '{u:10s}' | Senha: '{s:5s}' -> ACESSO NEGADO! Credenciais inválidas.")
```

### Análise e Conclusão (Exercício 13)

**a) Como funciona a tabela-verdade do operador `and`?**
A expressão `A and B` só retorna `True` quando **todas** as condições forem verdadeiras simultaneamente. Se qualquer uma falhar, o acesso é negado.

**b) O que garante a proteção das contas?**
Saber o nome do usuário não concede acesso, assim como ter a senha em uma conta incorreta também não autoriza o login.

**Conclusão:** A autenticação baseada em `and` assegura a consistência das credenciais, sendo o pilar dos fluxos de autorização em sistemas computacionais.

## Exercício 14 — Desconto comercial para compra expressiva ou cliente VIP (Operador OR) (pág. 34)
> **Enunciado:** "Solicite o valor de uma compra e informe se o cliente é VIP. O cliente receberá 10% de desconto quando a compra for superior a R$ 500 ou ele for VIP."

```python
# 1. FUNÇÃO DE BENEFÍCIO COMERCIAL
def aplicar_desconto_vip(valor_compra: float, is_vip: bool):
    # Condição: compra > 500 OU cliente VIP
    tem_desconto = (valor_compra > 500.00) or is_vip
    
    taxa_desconto = 0.10 if tem_desconto else 0.00
    desconto = valor_compra * taxa_desconto
    valor_pago = valor_compra - desconto
    
    return {
        "valor_compra": valor_compra,
        "is_vip": is_vip,
        "tem_desconto": tem_desconto,
        "taxa_pct": taxa_desconto * 100,
        "desconto": desconto,
        "valor_pago": valor_pago
    }

# 2. TESTES DE CENÁRIOS
pedidos = [
    (350.00, False),  # Baixo valor e Não-VIP (False or False -> Sem desconto)
    (750.00, False),  # Alto valor e Não-VIP (True or False -> Desconto)
    (200.00, True),   # Baixo valor e VIP (False or True -> Desconto)
    (900.00, True)    # Alto valor e VIP (True or True -> Desconto)
]

print("=== AVALIAÇÃO DE DESCONTO VIP / VALOR ===")
for v, vip in pedidos:
    r = aplicar_desconto_vip(v, vip)
    tipo_cliente = "VIP" if vip else "Comum"
    status_desc = f"Recebe 10% de desconto (-R$ {r['desconto']:.2f})" if r['tem_desconto'] else "Sem desconto"
    print(f"Compra: R$ {v:6.2f} | Tipo: {tipo_cliente:5s} -> {status_desc} | Total a pagar: R$ {r['valor_pago']:6.2f}")
```

### Análise e Conclusão (Exercício 14)

**a) Em quais situações o cliente recebe o desconto de 10%?**
Em três de quatro cenários possíveis: se comprar mais de R$ 500, se for VIP independente do valor, ou se cumprir ambas as condições.

**b) Qual a vantagem estratégica dessa regra para a empresa?**
Fideliza clientes VIP recorrentes mesmo em compras menores e estimula clientes comuns a elevarem o valor de seus pedidos acima de R$ 500.

**Conclusão:** O operador `or` viabiliza políticas de bonificação flexíveis, contemplando múltiplos caminhos de elegibilidade para os clientes.

## Exercício 15 — Monitoramento de temperatura e umidade crítica (Operador AND) (pág. 34)
> **Enunciado:** "Solicite a temperatura e a umidade de um ambiente industrial. Exiba um alerta quando a temperatura ultrapassar 35 °C e a umidade estiver abaixo de 30%."

```python
# 1. FUNÇÃO DE MONITORAMENTO AMBIENTAL
def monitorar_clima_industrial(temperatura_c: float, umidade_pct: float):
    # Condição crítica: temperatura > 35 E umidade < 30
    alerta_critico = (temperatura_c > 35.0) and (umidade_pct < 30.0)
    
    if alerta_critico:
        diagnostico = "ALERTA CRÍTICO: Ambiente quente e seco! Risco de sobreaquecimento e eletricidade estática."
    else:
        diagnostico = "Ambiente dentro dos parâmetros normais de trabalho."
        
    return {
        "temperatura": temperatura_c,
        "umidade": umidade_pct,
        "alerta": alerta_critico,
        "diagnostico": diagnostico
    }

# 2. LEITURAS EM DIFERENTES GALPÕES
leituras = [
    (28.0, 55.0),  # Agradável (False and False)
    (38.0, 45.0),  # Quente, mas úmido (True and False)
    (30.0, 22.0),  # Frio/ameno, mas seco (False and True)
    (39.5, 18.0)   # Quente e muito seco (True and True -> ALERTA)
]

print("=== CONTROLE METEOROLÓGICO INDUSTRIAL (AND) ===")
for t, u in leituras:
    res = monitorar_clima_industrial(t, u)
    simbolo = "🔥 ALERTA" if res['alerta'] else "✅ NORMAL"
    print(f"Temp: {t:4.1f} °C | Umidade: {u:4.1f}% | {simbolo:10s} -> {res['diagnostico']}")
```

### Análise e Conclusão (Exercício 15)

**a) Por que o operador `and` é exigido neste contexto?**
Porque uma temperatura de 36 °C com 60% de umidade não gera o mesmo risco eletrostático que um ambiente com apenas 20% de umidade. O alerta só deve soar quando ambas as condições de risco coexistirem.

**b) Qual ação corretiva deve ser tomada quando o alerta disparar?**
Acionamento de sistemas de umidificação industrial e ajuste dos condicionadores de ar para restabelecer a faixa segura de umidade (> 40%) e temperatura (< 25 °C).

**Conclusão:** A verificação com `and` assegura diagnósticos de alta especificidade, prevenindo falsos alarmes em sistemas de controle predial e fabril.
