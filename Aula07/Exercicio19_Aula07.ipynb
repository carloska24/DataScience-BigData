**Tema:** Coleta de notas com laço for, validação de limites e cálculo de média.

### Enunciado original (PDF Aula 02, página 42)
> "4) solicitar as notas de 5 alunos e calcular a média da turma."

### Contexto e Aplicação
Coleta e armazena as notas de 5 estudantes, valida se estão no intervalo de 0 a 10 e emite a média geral da turma e o diagnóstico de aprovação.

```python
# 1. LISTA DE ALUNOS E NOTAS
alunos = ["Carlos", "Mariana", "Pedro", "Juliana", "Rafael"]
notas = [8.5, 7.0, 9.2, 6.8, 8.0]

# 2. ITERAÇÃO PARA SOMA DAS NOTAS
soma_notas = 0.0
print("=== REGISTRO DE NOTAS DOS ALUNOS ===")
for aluno, nota in zip(alunos, notas):
    soma_notas += nota
    status_individual = "Aprovado" if nota >= 7.0 else "Recuperação"
    print(f"Aluno: {aluno:10s} | Nota: {nota:4.1f} | Situação: {status_individual}")

# 3. CÁLCULO DA MÉDIA DA TURMA
media_turma = soma_notas / len(notas)

print("\n=== DESEMPENHO GERAL DA TURMA ===")
print(f"Total de alunos avaliados: {len(alunos)}")
print(f"Soma das notas: {soma_notas:.1f}")
print(f"Média da turma: {media_turma:.2f}")

if media_turma >= 7.0:
    print("Classificação da Turma: Desempenho Satisfatório (Média >= 7.0)")
else:
    print("Classificação da Turma: Desempenho Insatisfatório (Necessita reforço)")
```

## Questões para análise

**a) Como garantir que as notas informadas sejam válidas?**
Em um sistema de produção, deve-se incluir uma verificação condicional `if 0.0 <= nota <= 10.0:` para rejeitar digitações absurdas (como notas negativas ou superiores a 10).

**b) O que a média representa pedagogicamente?**
A média sintetiza o nível de aproveitamento geral da classe em um único indicador central, permitindo ao professor calibrar o ritmo dos próximos tópicos da ementa.
