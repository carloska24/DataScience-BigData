**Tema:** Avaliação de indicadores de desempenho fabril (KPIs) com estruturas condicionais.

### Enunciado original (PDF Aula 02, página 26)
> "Escreva um programa que solicite o percentual de crescimento de produção de uma empresa e informe se houve um crescimento (porcentagem positiva) ou um decrescimento (porcentagem negativa)."

### Contexto e Aplicação
Na gestão industrial, monitorar se a linha de montagem superou ou ficou aquém da meta do período anterior é fundamental para tomada de decisões operacionais. O código analisa o sinal da taxa percentual.

```python
# 1. FUNÇÃO DE CLASSIFICAÇÃO DA PRODUÇÃO
def avaliar_crescimento(percentual: float) -> str:
    """Classifica a variação da produção conforme o sinal do percentual."""
    if percentual > 0:
        return f"Crescimento na produção: aumento de +{percentual:.2f}%."
    elif percentual < 0:
        return f"Decrescimento na produção: queda de {percentual:.2f}%."
    else:
        return "Estabilidade na produção: variação nula (0.00%)."

# 2. TESTES COM DADOS INDUSTRIAIS HISTÓRICOS
historico_meses = {
    "Mês": ["Janeiro", "Fevereiro", "Março", "Abril", "Maio"],
    "Percentual (%)": [12.5, -4.8, 0.0, 8.3, -15.2]
}

print("=== ANÁLISE DE HISTÓRICO DE PRODUÇÃO ===")
for mes, pct in zip(historico_meses["Mês"], historico_meses["Percentual (%)"]):
    resultado = avaliar_crescimento(pct)
    print(f"{mes:10s} | Taxa: {pct:+6.2f}% -> {resultado}")
```

## Questões para análise

**a) Como o sistema diferencia crescimento de decrescimento?**
Através da verificação lógica do sinal do valor: valores estritamente maiores que 0 (`> 0`) representam crescimento; valores estritamente menores que 0 (`< 0`) representam decrescimento; e o valor 0 representa estabilidade.

**b) Por que é recomendável prever o caso neutro (igual a zero)?**
Se houvesse apenas `if percentual > 0: ... else: ...`, um valor de 0% seria reportado erroneamente como decrescimento. A inclusão do caso zero garante precisão técnica ao relatório.

## Conclusão

A análise condicional do percentual de crescimento permite que a diretoria fabril receba alertas automáticos e relatórios consistentes sobre a expansão ou retração das unidades produtivas.
