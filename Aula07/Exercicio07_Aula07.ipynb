### Enunciado original (PDF Aula 02, página 29)
> "Desenvolva um programa que solicite dois números e uma operação matemática (+, -, * ou /). Utilize estruturas condicionais para identificar a operação e apresentar o resultado. Caso o usuário escolha uma divisão por zero, exiba uma mensagem de erro."

### Contexto e Aplicação
Executa as quatro operações elementares com validação explícita de divisor nulo para prevenir interrupção abrupta do programa.

```python
# 1. FUNÇÃO DA CALCULADORA
def calcular(num1: float, num2: float, operacao: str):
    operacao = operacao.strip()
    
    if operacao == "+":
        return num1 + num2, None
    elif operacao == "-":
        return num1 - num2, None
    elif operacao == "*":
        return num1 * num2, None
    elif operacao == "/":
        if num2 == 0:
            return None, "ERRO: Divisão por zero não é permitida!"
        return num1 / num2, None
    else:
        return None, f"ERRO: Operação '{operacao}' inválida! Escolha +, -, * ou /."

# 2. TESTES DE TODAS AS OPERAÇÕES E CASOS ESPECIAIS
testes = [
    (15, 3, "+"),
    (20, 7, "-"),
    (6, 8, "*"),
    (48, 6, "/"),
    (10, 0, "/"),  # Divisão por zero
    (5, 2, "%")    # Operador desconhecido
]

print("=== DEMONSTRAÇÃO DA CALCULADORA ===")
for n1, n2, op in testes:
    res, err = calcular(n1, n2, op)
    if err:
        print(f"{n1} {op} {n2} -> {err}")
    else:
        print(f"{n1} {op} {n2} = {res:.2f}")
```

## Questões para análise

**a) Por que o teste do divisor zero precisa ocorrer antes da divisão?**
Em Python, tentar calcular `x / 0` gera uma exceção `ZeroDivisionError` que interrompe o programa. Ao verificar com `if num2 == 0:` antes da operação, tratamos a falha de forma graciosa e amigável.

**b) Quais operadores são aceitos?**
Adição (`+`), subtração (`-`), multiplicação (`*`) e divisão (`/`). Operadores não previstos acionam o bloco final de erro de sintaxe.

## Conclusão

A calculadora condicional demonstra a implementação correta de seletores de operações e a prevenção de erros críticos de divisão por zero.
