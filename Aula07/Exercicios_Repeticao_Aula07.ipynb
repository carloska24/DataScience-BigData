## Exercício 16 — Contagem sequencial de 1 a 100 com estrutura while (pág. 42)
> **Enunciado:** "1) Mostrar os números de 1 a 100 utilizando while"

```python
# 1. INICIALIZAÇÃO DA VARIÁVEL DE CONTROLE
contador = 1
numeros = []

# 2. ESTRUTURA WHILE DE REPETIÇÃO
while contador <= 100:
    numeros.append(contador)
    contador += 1  # Incremento fundamental para encerramento do laço

# 3. EXIBIÇÃO FORMATADA DOS NÚMEROS (EM BLOCOS DE 10 PARA LEITURA CLARA)
print("=== CONTAGEM DE 1 A 100 COM WHILE ===")
for i in range(0, 100, 10):
    linha = " ".join(f"{num:3d}" for num in numeros[i:i+10])
    print(linha)

print(f"\nContador final após o loop: {contador}")
print(f"Total de números gerados: {len(numeros)}")
```

### Análise e Conclusão (Exercício 16)

**a) Quais são os três elementos fundamentais de um laço `while`?**
1. Inicialização da variável de controle (`contador = 1`).
2. Condição de continuidade (`contador <= 100`).
3. Atualização/incremento da variável a cada iteração (`contador += 1`).

**b) O que aconteceria se a linha `contador += 1` fosse omitida?**
O valor de `contador` permaneceria eternamente igual a 1, a condição `1 <= 100` seria sempre verdadeira e o programa entraria em um laço infinito, consumindo recursos e travando o processo.

**Conclusão:** O laço `while` executa repetições com precisão desde que sua condição de parada e seu incremento estejam estritamente coordenados.

## Exercício 17 — Tabuada interativa de 1 a 10 com laço for (pág. 42)
> **Enunciado:** "2) Solicitar um número e apresentar sua tabuada de 1 a 10 utilizando for."

```python
# 1. FUNÇÃO DE GERAÇÃO DE TABUADA
def gerar_tabuada(numero: int):
    print(f"=== TABUADA DO NÚMERO {numero} ===")
    tabela = []
    for i in range(1, 11):
        resultado = numero * i
        tabela.append((i, resultado))
        print(f"{numero:2d} x {i:2d} = {resultado:3d}")
    return tabela

# 2. DEMONSTRAÇÃO COM NÚMEROS DE EXEMPLO
numero_escolhido = 7
tabuada_7 = gerar_tabuada(numero_escolhido)

# 3. TABUADA ADICIONAL (EX: BASE INDUSTRIAL / CÁLCULO DE PEÇAS)
print("\n=== APLICAÇÃO INDUSTRIAL: CÁLCULO DE LOTES (FATOR 12) ===")
gerar_tabuada(12)
```

### Análise e Conclusão (Exercício 17)

**a) Como funciona o `range(1, 11)` no Python?**
A função `range(início, fim)` gera valores a partir de `início` (inclusivo) até `fim - 1` (exclusivo). Assim, `range(1, 11)` produz exatamente a sequência de 1 a 10.

**b) Por que o `for` é preferível ao `while` para tabuadas?**
Porque o número de repetições é conhecido a priori (exatamente 10 iterações), tornando o `for` mais conciso, legível e imune a erros de esquecimento de incremento.

**Conclusão:** O laço `for` com `range()` é a ferramenta padrão e mais elegante do Python para iterações sobre intervalos matemáticos e contagens pré-determinadas.

## Exercício 18 — Acumulador numérico e somatório de 10 valores (pág. 42)
> **Enunciado:** "3) solicitar 10 números e calcular a soma total."

```python
# 1. DADOS DE ENTRADA (10 VALORES COLETADOS)
valores = [15.2, 22.8, 18.0, 35.4, 42.1, 19.5, 27.3, 31.0, 14.9, 25.8]

# 2. CÁLCULO DA SOMA COM LAÇO FOR E ACUMULADOR
soma_total = 0.0
print("=== PROCESSAMENTO DOS 10 NÚMEROS ===")
for i, valor in enumerate(valores, start=1):
    soma_total += valor
    print(f"Passo {i:2d} | Valor recebido: {valor:5.1f} | Soma parcial: {soma_total:6.1f}")

media_valores = soma_total / len(valores)

# 3. RELATÓRIO ESTATÍSTICO
print("\n=== RESULTADO FINAL ===")
print(f"Quantidade de números processados: {len(valores)}")
print(f"Soma total acumulada: {soma_total:.2f}")
print(f"Média aritmética dos valores: {media_valores:.2f}")
print(f"Maior valor da série: {max(valores):.2f}")
print(f"Menor valor da série: {min(valores):.2f}")
```

### Análise e Conclusão (Exercício 18)

**a) O que caracteriza uma variável acumuladora?**
É uma variável inicializada em zero (`soma_total = 0.0`) que, a cada ciclo do laço, recebe o valor atual somado ao seu próprio conteúdo anterior (`soma_total += valor`).

**b) O resultado pode ser validado pela função `sum()` nativa?**
Sim, `sum(valores)` retorna exatamente a mesma soma, comprovando que o algoritmo manual implementado reproduz fielmente as operações vetorizadas de bibliotecas como NumPy e Pandas.

**Conclusão:** A técnica de acumulação por iteração é a fundação para cálculos de somatórios, integrais numéricas e agregações de dados em Data Science.

## Exercício 19 — Cálculo da média da turma com repetição e notas de alunos (pág. 42)
> **Enunciado:** "4) solicitar as notas de 5 alunos e calcular a média da turma."

```python
# 1. LISTA DE ALUNOS E NOTAS
alunos = ["Carlos", "Mariana", "Pedro", "Juliana", "Rafael"]
notas = [8.5, 7.0, 9.2, 6.8, 8.0]

# 2. ITERAÇÃO PARA SOMA DAS NOTAS
soma_notas = 0.0
print("=== REGISTRO DE NOTAS DOS ALUNOS ===")
for aluno, nota in zip(alunos, notas):
    soma_notas += nota
    status_individual = "Aprovado" if nota >= 7.0 else "Recuperação"
    print(f"Aluno: {aluno:10s} | Nota: {nota:4.1f} | Situação: {status_individual}")

# 3. CÁLCULO DA MÉDIA DA TURMA
media_turma = soma_notas / len(notas)

print("\n=== DESEMPENHO GERAL DA TURMA ===")
print(f"Total de alunos avaliados: {len(alunos)}")
print(f"Soma das notas: {soma_notas:.1f}")
print(f"Média da turma: {media_turma:.2f}")

if media_turma >= 7.0:
    print("Classificação da Turma: Desempenho Satisfatório (Média >= 7.0)")
else:
    print("Classificação da Turma: Desempenho Insatisfatório (Necessita reforço)")
```

### Análise e Conclusão (Exercício 19)

**a) Como garantir que as notas informadas sejam válidas?**
Em um sistema de produção, deve-se incluir uma verificação condicional `if 0.0 <= nota <= 10.0:` para rejeitar digitações absurdas (como notas negativas ou superiores a 10).

**b) O que a média representa pedagogicamente?**
A média sintetiza o nível de aproveitamento geral da classe em um único indicador central, permitindo ao professor calibrar o ritmo dos próximos tópicos da ementa.

**Conclusão:** A estrutura de repetição automatiza o processamento de avaliações acadêmicas e corporativas, viabilizando análises de desempenho coletivo.

## Exercício 20 — Inspeção térmica de 10 máquinas: contagem em alerta e média (pág. 42)
> **Enunciado:** "5) solicitar a temperatura de 10 máquinas, contar quantas ultrapassaram 80 °C e calcular a temperatura média."

```python
# 1. LEITURA DE 10 MÁQUINAS INDUSTRIAIS
dados_inspecao = [
    ("Torno CNC 01", 68.5),
    ("Fresa 02", 74.0),
    ("Prensa Hidráulica 03", 85.0), # Alerta
    ("Compressor A 04", 62.0),
    ("Injetora 05", 91.5),          # Alerta
    ("Retificadora 06", 58.0),
    ("Centro de Usinagem 07", 83.2), # Alerta
    ("Torno Convencional 08", 45.0),
    ("Esteira Transportadora 09", 52.0),
    ("Solda Robotizada 10", 88.0)   # Alerta
]

# 2. PROCESSAMENTO COM LAÇO FOR
soma_temperaturas = 0.0
maquinas_em_alerta = 0
lista_alertas = []

print("=== INSPEÇÃO TÉRMICA DE 10 MÁQUINAS INDUSTRIAIS ===")
for i, (nome, temp) in enumerate(dados_inspecao, start=1):
    soma_temperaturas += temp
    em_alerta = temp > 80.0
    if em_alerta:
        maquinas_em_alerta += 1
        lista_alertas.append((nome, temp))
        aviso = "🔥 ALERTA (> 80 °C)"
    else:
        aviso = "✅ NORMAL"
    print(f"Máquina {i:2d}: {nome:25s} | Temperatura: {temp:5.1f} °C | Status: {aviso}")

# 3. CÁLCULO DA MÉDIA E RELATÓRIO CONSOLIDADO
media_temperatura = soma_temperaturas / len(dados_inspecao)

print("\n=== RELATÓRIO CONSOLIDADO DA MANUTENÇÃO ===")
print(f"Total de máquinas inspecionadas: {len(dados_inspecao)}")
print(f"Temperatura média do parque: {media_temperatura:.1f} °C")
print(f"Quantidade de máquinas acima de 80 °C: {maquinas_em_alerta} ({maquinas_em_alerta/len(dados_inspecao)*100:.0f}%)")
print("\nMáquinas que necessitam de intervenção imediata:")
for m, t in lista_alertas:
    print(f" -> {m} com {t:.1f} °C")
```

### Análise e Conclusão (Exercício 20)

**a) Como o contador e o acumulador atuam em conjunto no laço?**
Enquanto o acumulador (`soma_temperaturas += temp`) soma incondicionalmente todos os valores para permitir o cálculo da média final, o contador (`maquinas_em_alerta += 1`) só incrementa quando a condição `temp > 80.0` é satisfeita.

**b) Qual a relevância desse algoritmo para a Indústria 4.0?**
Em plantas conectadas por IoT (Internet das Coisas), algoritmos como este processam streams de dados de centenas de sensores a cada segundo, disparando ordens de serviço automáticas para o setor de manutenção preventiva.

**Conclusão:** A combinação de laços com acumuladores e filtros condicionais constitui a base da telemetria e do monitoramento em tempo real na indústria inteligente.
