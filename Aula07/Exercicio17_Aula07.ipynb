### Contexto e Aplicação
Gera a tabuada de multiplicação de um número escolhido no intervalo de 1 a 10, com formatação tabular e visualização limpa.

```python
# 1. FUNÇÃO DE GERAÇÃO DE TABUADA
def gerar_tabuada(numero: int):
    print(f"=== TABUADA DO NÚMERO {numero} ===")
    tabela = []
    for i in range(1, 11):
        resultado = numero * i
        tabela.append((i, resultado))
        print(f"{numero:2d} x {i:2d} = {resultado:3d}")
    return tabela

# 2. DEMONSTRAÇÃO COM NÚMEROS DE EXEMPLO
numero_escolhido = 7
tabuada_7 = gerar_tabuada(numero_escolhido)

# 3. TABUADA ADICIONAL (EX: BASE INDUSTRIAL / CÁLCULO DE PEÇAS)
print("\n=== APLICAÇÃO INDUSTRIAL: CÁLCULO DE LOTES (FATOR 12) ===")
gerar_tabuada(12)
```

## Questões para análise

**a) Como funciona o `range(1, 11)` no Python?**
A função `range(início, fim)` gera valores a partir de `início` (inclusivo) até `fim - 1` (exclusivo). Assim, `range(1, 11)` produz exatamente a sequência de 1 a 10.

**b) Por que o `for` é preferível ao `while` para tabuadas?**
Porque o número de repetições é conhecido a priori (exatamente 10 iterações), tornando o `for` mais conciso, legível e imune a erros de esquecimento de incremento.

## Conclusão

O laço `for` com `range()` é a ferramenta padrão e mais elegante do Python para iterações sobre intervalos matemáticos e contagens pré-determinadas.
