## Exercício 1 — Produção e consumo de energia

```python
import pandas as pd
from IPython.display import display

dados = {
    "maquina": ["A", "B", "C", "D", "E"],
    "producao": [1250, 1380, 1190, 1450, 1320],
    "energia": [420, 465, 390, 510, 405]
}
maquinas = pd.DataFrame(dados)
display(maquinas)
```

### a) Qual máquina produziu mais?

```python
display(maquinas.loc[maquinas["producao"].idxmax(), ["maquina", "producao"]])
```

A máquina D produziu mais, com 1.450 peças por turno.

### b) Qual máquina consome menos energia?

```python
display(maquinas.loc[maquinas["energia"].idxmin(), ["maquina", "energia"]])
```

A máquina C consumiu menos energia: 390 kWh por turno.

### c) Qual foi a produção média das máquinas?

```python
print(maquinas["producao"].mean())
```

A produção média foi de 1.318 peças por turno.

### d) Qual foi o consumo médio de energia?

```python
print(maquinas["energia"].mean())
```

O consumo médio foi de 438 kWh por turno.

### e) Ordene as máquinas da maior para a menor produção.

```python
display(maquinas.sort_values("producao", ascending=False))
```

Da maior para a menor produção: D, B, E, A e C.

### f) Crie a coluna eficiência, calculada por produção/energia.

```python
maquinas["eficiencia"] = maquinas["producao"] / maquinas["energia"]
display(maquinas.round(4))
```

A eficiência foi calculada dividindo a produção pelo consumo de energia, em peças por kWh.

### g) Qual máquina apresentou a melhor relação entre produção e consumo de energia?

```python
display(maquinas.loc[maquinas["eficiencia"].idxmax(), ["maquina", "eficiencia"]])
```

A máquina E teve a melhor eficiência, com cerca de 3,26 peças por kWh.

## Exercício 2 — Viagens e distância percorrida

Base: [viagens_por_dia_100.csv](https://drive.google.com/file/d/1qW-44xXKdYB2WmbKNyd17O5t_xuE70ZX/view).

```python
url_viagens = (
    "https://drive.google.com/uc?export=download"
    "&id=1qW-44xXKdYB2WmbKNyd17O5t_xuE70ZX"
)

viagens = pd.read_csv(url_viagens)
viagens["data"] = pd.to_datetime(viagens["data"])
display(viagens.head())
```

### a) Quantas viagens foram realizadas durante todo o período analisado?

```python
print(viagens["quantidade_viagens"].sum())
```

Foram realizadas 2.893 viagens entre 01/01/2026 e 10/04/2026.

### b) Qual dia apresentou a maior quantidade de viagens?

```python
display(viagens.loc[viagens["quantidade_viagens"].idxmax(),
                   ["data", "dia_semana", "quantidade_viagens"]])
```

O dia com mais viagens foi 26/02/2026, uma quinta-feira, com 48 viagens.

### c) Qual é a média diária de viagens realizadas?

```python
print(viagens["quantidade_viagens"].mean())
```

A média foi de 28,93 viagens por dia.

### d) Em qual mês foi registrada a maior quantidade total de viagens?

```python
viagens["mes"] = viagens["data"].dt.to_period("M")
viagens_mes = viagens.groupby("mes")["quantidade_viagens"].sum()
display(viagens_mes)
print("Mês com mais viagens:", viagens_mes.idxmax())
```

Janeiro teve o maior total, com 901 viagens. Abril tem dados somente até o dia 10.

### e) Existe correlação entre a quantidade de viagens e a distância total percorrida?

```python
correlacao_viagens = viagens["quantidade_viagens"].corr(viagens["distancia_total_km"])
print(round(correlacao_viagens, 4))
```

Sim. A correlação foi de 0,8694, uma relação positiva forte. Em geral, os dias com mais viagens também tiveram maior distância percorrida.

## Exercício 3 — Vendas do Produto A

Base: [vendas_mensais_100.csv](https://drive.google.com/file/d/1K1ekAHpbd6MXyxODX0CnQkk3061qYEp1/view).

```python
url_vendas = (
    "https://drive.google.com/uc?export=download"
    "&id=1K1ekAHpbd6MXyxODX0CnQkk3061qYEp1"
)

vendas = pd.read_csv(url_vendas)
vendas["mes"] = pd.to_datetime(vendas["mes"])
display(vendas.head())
```

### a) Qual foi a quantidade total de unidades vendidas no período analisado?

```python
print(vendas["unidades_vendidas"].sum())
```

Foram vendidas 24.499 unidades do Produto A entre janeiro de 2018 e abril de 2026.

### b) Qual mês apresentou o maior faturamento?

```python
display(vendas.loc[vendas["faturamento_rs"].idxmax(), ["mes", "faturamento_rs"]])
```

O maior faturamento foi em novembro de 2025: R$ 22.516,32.

### c) Qual foi o faturamento total obtido com a comercialização do Produto A?

```python
print(round(vendas["faturamento_rs"].sum(), 2))
```

O faturamento total foi de R$ 1.281.779,17.

### d) Qual ano apresentou a maior quantidade de unidades vendidas?

```python
vendas["ano"] = vendas["mes"].dt.year
vendas_ano = vendas.groupby("ano")["unidades_vendidas"].sum()
display(vendas_ano)
print("Ano com mais vendas:", vendas_ano.idxmax())
```

O ano de 2025 teve mais vendas, com 3.511 unidades. Em 2026, a base tem dados apenas de janeiro a abril.

### e) Existe relação entre o preço unitário e a quantidade de unidades vendidas?

```python
correlacao_vendas = vendas["preco_unitario_rs"].corr(vendas["unidades_vendidas"])
print(round(correlacao_vendas, 4))
```

Sim. A correlação foi positiva, de 0,7713. Os preços e as quantidades vendidas aumentaram ao longo do período. Isso não prova que o aumento do preço causou mais vendas, pois os dois podem ter acompanhado o crescimento ao longo dos anos.
